```python
#Final R studio data following Mahalanobis' distance outlier removal 

import pandas as pd
import numpy as np
import scipy.stats as sc 
from scipy.stats import chi2
reshaped_dfgrouped_avg_chasetry = pd.read_csv('/Users/macbook/Desktop/7 UCL PhD (2022-2025)/2 Minecraft fNIRS-fMRI Study/Study Files/Minecraftanalysis/Analysispipeline/Behaviouraldata/FINALFRAMEFORMATRIXRSTUDIO.csv')
def calculateMahalanobis(y=None, data=None, cov=None):
  
    y_mu = y - np.mean(data)
    if not cov:
        cov = np.cov(data.values.T)
    inv_covmat = np.linalg.inv(cov)
    left = np.dot(y_mu, inv_covmat)
    mahal = np.dot(left, y_mu.T)
    return mahal.diagonal()

reshaped_dfgrouped_avg_chasetry['Hunting original'] = reshaped_dfgrouped_avg_chasetry['Hunting Success']
reshaped_dfgrouped_avg_chasetry['Hunting Success'] = sc.zscore(reshaped_dfgrouped_avg_chasetry['Hunting Success'])
reshaped_dfgrouped_avg_chasetry['Mahalanobis'] = calculateMahalanobis(y=reshaped_dfgrouped_avg_chasetry[['Distance between self and other', 'Distance between self and prey','Scaled self facing prey', 'Scaled self facing other','Lag Score','Hunting Success','Heart Rate self','Breathing Rate self']],data=reshaped_dfgrouped_avg_chasetry[['Distance between self and other', 'Distance between self and prey','Scaled self facing prey', 'Scaled self facing other','Lag Score','Hunting Success','Heart Rate self','Breathing Rate self']])
reshaped_dfgrouped_avg_chasetry['m_p'] = 1 - chi2.cdf(reshaped_dfgrouped_avg_chasetry['Mahalanobis'], 8)
malaxvariable=chi2.cdf(reshaped_dfgrouped_avg_chasetry['Mahalanobis'], 8)
malayvariable=reshaped_dfgrouped_avg_chasetry['Mahalanobis']
malayvariable=pd.DataFrame(malayvariable)
malayvariable = malayvariable.reset_index(drop=True)
malayvariable.columns=['malayvals']
malaxvariable=pd.DataFrame(malaxvariable)
malaxvariable.columns=['malaxvals']
malaxvariable = malaxvariable.reset_index(drop=True)
mal = pd.concat([malayvariable,malaxvariable],axis=1)
mal.sort_values(by='malaxvals',ascending=True)
mal = mal.round(1)
minval = mal[mal['malaxvals']==1.0].min()[0]
mal.to_csv('/Users/macbook/Desktop/7 UCL PhD (2022-2025)/2 Minecraft fNIRS-fMRI Study/Study Files/fMRIstudy/Minecraft fMRI Real Data/malvals.csv')
FINALDATAREMOVED2 = reshaped_dfgrouped_avg_chasetry[reshaped_dfgrouped_avg_chasetry['Mahalanobis']>minval]
FINALDATA2 = reshaped_dfgrouped_avg_chasetry[reshaped_dfgrouped_avg_chasetry['Mahalanobis']<=minval]
FINALDATA2.to_csv('/Users/macbook/Desktop/7 UCL PhD (2022-2025)/2 Minecraft fNIRS-fMRI Study/Study Files/Minecraftanalysis/Analysispipeline/Behaviouraldata/FINALFRAMEFORMATRIXRSTUDIOMAHAL.csv')
```
