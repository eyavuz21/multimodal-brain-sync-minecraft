```python
import cv2
import numpy as np

input_brain_path = '/Users/macbook/Desktop/7 UCL PhD (2022-2025)/2 Minecraft fNIRS-fMRI Study/Study Files/Minecraftanalysis/Analysispipeline/fNIRSanalysispipeline/Plottinganalysingfnirsdatascripts/inputbrain.png'
brain_template_path = '/Users/macbook/Desktop/7 UCL PhD (2022-2025)/2 Minecraft fNIRS-fMRI Study/Study Files/Minecraftanalysis/Analysispipeline/fNIRSanalysispipeline/Plottinganalysingfnirsdatascripts/Braintemplate.png'
input_brain = cv2.imread(input_brain_path)
brain_template = cv2.imread(brain_template_path)
brain_template = cv2.resize(brain_template, (input_brain.shape[1], input_brain.shape[0]))
circle_count = 0
circles = []
template_center_set = False
template_center = (0, 0)
current_color = (0,0,255)
def draw_circle_on_input_brain(event, x, y, flags, param):
    global input_brain, circles, circle_count
    if event == cv2.EVENT_LBUTTONDOWN:  
        circle_count += 1
        color = (255,0,0)
        cv2.circle(input_brain, (x, y), radius=12, color=color, thickness=-1)
        cv2.putText(input_brain, str(circle_count), (x-7, y+5), cv2.FONT_HERSHEY_SIMPLEX, 0.5, (255,255,255), 2)
        circles.append((x, y, color, circle_count))
        cv2.imshow('Input Brain with Circles', input_brain)
    elif event == cv2.EVENT_RBUTTONDOWN:  
        circle_count += 1
        color = current_color
        cv2.circle(input_brain, (x, y), radius=12, color=color, thickness=-1)
        cv2.putText(input_brain, str(circle_count), (x-7, y+5), cv2.FONT_HERSHEY_SIMPLEX, 0.5, (255,255,255), 2)
        circles.append((x, y, color, circle_count))
        cv2.imshow('Input Brain with Circles', input_brain)
    elif event == cv2.EVENT_LBUTTONDBLCLK:
        circle_count += 1
        color = (0,255,0)
        cv2.circle(input_brain, (x, y), radius=12, color=color, thickness=-1)
        cv2.putText(input_brain, str(circle_count), (x-7, y+5), cv2.FONT_HERSHEY_SIMPLEX, 0.5, (255,255,255), 2)
        circles.append((x, y, color, circle_count))
        cv2.imshow('Input Brain with Circles', input_brain)
def set_template_center(event, x, y, flags, param):
    global template_center, template_center_set
    if event == cv2.EVENT_LBUTTONDOWN or event == cv2.EVENT_RBUTTONDOWN: 
        template_center = (x, y)
        template_center_set = True
        print(f"Template center set at {template_center}")
        cv2.circle(brain_template, template_center, radius=12, color=(255,255,255), thickness=-1)
        cv2.imshow('Brain Template with Circles', brain_template)
cv2.namedWindow('Input Brain with Circles')
cv2.setMouseCallback('Input Brain with Circles', draw_circle_on_input_brain)
cv2.imshow('Input Brain with Circles', input_brain)
print("Draw grey dots with left-click and yellow dots with right-click on the input brain. Press ESC when done.")
while True:
    key = cv2.waitKey(1) & 0xFF
    if key == 27: 
        break
    elif key == ord('g'):  
        current_color = (0, 255, 0)
    elif key == ord('r'):  
        current_color = (0, 0, 255)
cv2.destroyAllWindows()
if len(circles) >= 3:
    print(f"Circles placed: {len(circles)}")
    print("Now, click on the center of the brain template.")
    cv2.namedWindow('Brain Template with Circles')
    cv2.setMouseCallback('Brain Template with Circles', set_template_center)
    cv2.imshow('Brain Template with Circles', brain_template)
    while not template_center_set:
        cv2.waitKey(1)
    cv2.destroyAllWindows()
    third_circle_position = circles[2]  
    print(f"3rd circle position on input brain: {third_circle_position[:2]}")
    offset_x = third_circle_position[0] - template_center[0]
    offset_y = third_circle_position[1] - template_center[1]
    alignment_offset = (offset_x, offset_y)
    print(f"Calculated alignment offset: {alignment_offset}")
    adjusted_circles = [(x - alignment_offset[0], y - alignment_offset[1], color, count) 
                        for (x, y, color, count) in circles]
    brain_template_with_circles = brain_template.copy()  
    for circle in adjusted_circles:
        cv2.circle(brain_template_with_circles, (int(circle[0]), int(circle[1])), radius=12, color=circle[2], thickness=-1)
        cv2.putText(brain_template_with_circles, str(circle[3]), (int(circle[0])-7, int(circle[1])+5), 
                    cv2.FONT_HERSHEY_SIMPLEX, 0.5, (255,255,255), 2)
    cv2.imshow('Brain Template with Adjusted Circles', brain_template_with_circles)
    cv2.waitKey(0)
    cv2.destroyAllWindows()
    output_path = '/Users/macbook/Desktop/7 UCL PhD (2022-2025)/2 Minecraft fNIRS-fMRI Study/Study Files/Minecraftanalysis/Analysispipeline/fNIRSanalysispipeline/Plottinganalysingfnirsdatascripts/FINALIMAGE.png'
    cv2.imwrite(output_path, brain_template_with_circles)
    print(f"Brain template with adjusted circles saved to {output_path}")
else:
    print("You need to draw at least 3 circles on the input brain.")
```
